```python
import pandas as pd 
import matplotlib.pyplot as plt

main_dir = "bench-output"

```

```python
# insert-random plot

df = pd.read_csv("bench-output/insert-random.csv")
df["Size"] = df["Name"].str.split("/").str[1]
df["DS"] = df["Name"].str.split("/").str[2]
df_g = df.groupby("DS")

rbt = df_g["RBT"]
bst = df_g["BST"]
x1 = rbt["Size"].to_numpy()
y1 = rbt["Mean"].to_numpy()

x2 = bst["Size"].to_numpy()
y2 = bst["Mean"].to_numpy()

plt.figure(figsize=(20, 10))
plt.title(f"Plot of random int insertions")
plt.xlabel("Input Size")
plt.ylabel("Average time (s)")
plt.xscale("log", base=2)

plt.plot(x1, y1, marker="o",color="blue")
plt.plot(x2, y2, marker="o",color="red")
plt.grid(True, which="both", linestyle=":")

# plt.yscale("log")

plt.legend()
plt.tight_layout()
plt.show()

print(rbt)
```
